```python
#13. Write a function grade(score) that returns:

#a. 90 and above: "Distinction"

#b. 75 to 89: "First Class"

#c. 60 to 74: "Second Class"

#d. below 60: "Fail"
#Apply it to scores and store the result in grades. (5)

def grade(scores):
    if scores>=90:
        return 'Distinction'
    elif scores>=75:
        return 'First class'
    elif scores>=60:
        return 'Second class'
    else:
        return 'Fail'
res=scores.apply(grade)
print(res)
```

```python
#14. Display how many students fall in each grade, then the same as percentages.
print(res.value_counts())
print(res.value_counts(normalize=True))
```

```python
#15. Display the unique grades and the number of unique grades.
a=res.unique()
print(f'unique grades:{a}')
print(f'number of unique grades:{len(a)}')
```
